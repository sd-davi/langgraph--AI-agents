### `add_messages` por dentro

`add_messages` é só uma função: dá para chamá-la diretamente e ver o que ela faz.

Preparando historico

```python
from langchain_core.messages import AIMessage, HumanMessage, RemoveMessage
from langgraph.graph.message import REMOVE_ALL_MESSAGES, add_messages
```

```python
historico = [
    HumanMessage("Qual o prazo?", id="1"),
    AIMessage("5 dias", id="2")
]
```

Aqui temos duas mensagens:
- HumanMessage: mensagem enviada pelo usuário.
- AIMessage: resposta do assistente.
O parâmetro id identifica cada mensagem individualmente.

```python
for m in historico:
    print(m)
```

```python
def mostarMensagens(*msg):
    return [f'{m.id} - {m.type} - {m.content}' for m in msg]
```

Essa função recebe várias mensagens usando *msgs e retorna uma lista de strings com três informações:
- ID da mensagem;
- Tipo da mensagem;
- Conteúdo.

#### As 4 operações do `add_messages`

##### 1 — Adicionar uma mensagem

```python
add_messages(historico,[('user','e o frete?')])
```

##### 2 — Substituir uma mensagem pelo mesmo ID

```python
add_messages(
    historico,
    [AIMessage("Até 5 dias úteis", id="2")]
)
```

Aqui está uma característica muito importante: se uma mensagem nova tiver o mesmo ID de uma mensagem existente, ela substitui a antiga, em vez de simplesmente ser adicionada.

##### 3 — Remover uma mensagem

```python
add_messages(historico, [RemoveMessage(id="1")])
```

O RemoveMessage informa ao reducer que a mensagem com aquele ID deve ser removida.

Nesse caso, a pergunta "Qual o prazo?" desaparece do histórico.

##### 4 — Remover todas as mensagens

```python
add_messages(
    historico,
    [
        RemoveMessage(id=REMOVE_ALL_MESSAGES),
        #HumanMessage("Nova conversa", id="9")
    ]
)
```

Aqui temos duas operações:
1. REMOVE_ALL_MESSAGES solicita a remoção de todas as mensagens existentes.
2. Uma nova mensagem é adicionada.
